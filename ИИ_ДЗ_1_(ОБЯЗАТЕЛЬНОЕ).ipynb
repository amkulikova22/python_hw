```python
import pandas as pd
import numpy as np

# Реальные средние температуры (примерные данные) для городов по сезонам
seasonal_temperatures = {
    "New York": {"winter": 0, "spring": 10, "summer": 25, "autumn": 15},
    "London": {"winter": 5, "spring": 11, "summer": 18, "autumn": 12},
    "Paris": {"winter": 4, "spring": 12, "summer": 20, "autumn": 13},
    "Tokyo": {"winter": 6, "spring": 15, "summer": 27, "autumn": 18},
    "Moscow": {"winter": -10, "spring": 5, "summer": 18, "autumn": 8},
    "Sydney": {"winter": 12, "spring": 18, "summer": 25, "autumn": 20},
    "Berlin": {"winter": 0, "spring": 10, "summer": 20, "autumn": 11},
    "Beijing": {"winter": -2, "spring": 13, "summer": 27, "autumn": 16},
    "Rio de Janeiro": {"winter": 20, "spring": 25, "summer": 30, "autumn": 25},
    "Dubai": {"winter": 20, "spring": 30, "summer": 40, "autumn": 30},
    "Los Angeles": {"winter": 15, "spring": 18, "summer": 25, "autumn": 20},
    "Singapore": {"winter": 27, "spring": 28, "summer": 28, "autumn": 27},
    "Mumbai": {"winter": 25, "spring": 30, "summer": 35, "autumn": 30},
    "Cairo": {"winter": 15, "spring": 25, "summer": 35, "autumn": 25},
    "Mexico City": {"winter": 12, "spring": 18, "summer": 20, "autumn": 15},
}

# Сопоставление месяцев с сезонами
month_to_season = {12: "winter", 1: "winter", 2: "winter",
                   3: "spring", 4: "spring", 5: "spring",
                   6: "summer", 7: "summer", 8: "summer",
                   9: "autumn", 10: "autumn", 11: "autumn"}

# Генерация данных о температуре
def generate_realistic_temperature_data(cities, num_years=10):
    dates = pd.date_range(start="2010-01-01", periods=365 * num_years, freq="D")
    data = []

    for city in cities:
        for date in dates:
            season = month_to_season[date.month]
            mean_temp = seasonal_temperatures[city][season]
            # Добавляем случайное отклонение
            temperature = np.random.normal(loc=mean_temp, scale=5)
            data.append({"city": city, "timestamp": date, "temperature": temperature})

    df = pd.DataFrame(data)
    df['season'] = df['timestamp'].dt.month.map(lambda x: month_to_season[x])
    return df

# Генерация данных
data = generate_realistic_temperature_data(list(seasonal_temperatures.keys()))
data.to_csv('temperature_data.csv', index=False)
```

```python
import asyncio
import time

# функция для подсчета скользящего среднего
async def calc_mean(data):
    data = data.set_index('timestamp').sort_index()
    data['rolling_temperature'] = data['temperature'].rolling('30D').mean()
    return data

# функция для подсчета cредней температуры и стандартное отклонение в разрезе города и сезона года, а так же для выявления аномалий
async def calc_std(data):
    avg_temp = data.groupby(['city', 'season']).agg(avg_temp=('temperature', 'mean'))
    temp_std= data.groupby(['city', 'season']).agg(temp_std=('temperature', 'std'))
    data = data.merge(avg_temp, on=['city', 'season'], how='left')
    data = data.merge(temp_std, on=['city', 'season'], how='left')
    data['upper_bound'] = data['avg_temp'] + 2 * data['temp_std']
    data['lower_bound'] = data['avg_temp'] - 2 * data['temp_std']
    data['is_anomaly'] = (data['temperature'] > data['upper_bound']) | (data['temperature'] < data['lower_bound'])
    return data

async def main():
    start_time = time.time()
    task1 = asyncio.create_task(calc_mean(data))
    task2 = asyncio.create_task(calc_std(data))

    # print(f"started at {time.strftime('%X')}")
    await task1
    await task2
    end_time = time.time()
    # print(f"finished at {time.strftime('%X')}")
    print((end_time-start_time)*1000)

final_result = await main()
```

### Текущая температура в Москве

```python
import requests

def get_temperature(city, api_key):
    url = f"https://api.openweathermap.org/data/2.5/weather"
    params = {
        'q': city,
        'appid': api_key,
        'units': 'metric',
        'lang': 'ru'
    }
    
    response = requests.get(url, params=params)
    data = response.json()
    
    if response.status_code == 200:
        temp = data['main']['temp']
        print(f"Температура в {city}: {temp}°C")
        return temp
    else:
        print(f"Ошибка: {data.get('message', 'Неизвестная ошибка')}")
        return None

# Использование
api_key = "13e489d16133d222e13e602f3c726e5e"
get_temperature("Москва", api_key)
```

## Синхронные и асинхронные методы

```python
%%time
import requests

def get_temperature(city, api_key):
    url = f"https://api.openweathermap.org/data/2.5/weather"
    params = {
        'q': city,
        'appid': api_key,
        'units': 'metric',
        'lang': 'ru'
    }
    
    response = requests.get(url, params=params)
    data = response.json()
    
    if response.status_code == 200:
        temp = data['main']['temp']
        print(f"Температура в {city}: {temp}°C")
        return temp
    else:
        print(f"Ошибка: {data.get('message', 'Неизвестная ошибка')}")
        return None

# Использование
api_key = "13e489d16133d222e13e602f3c726e5e"
get_temperature("Москва", api_key)
get_temperature("Дубай", api_key)
get_temperature("Каир", api_key)
get_temperature("Берлин", api_key)
get_temperature("Пекин", api_key)
```

```python
import asyncio
import time
import requests

async def get_temperature(city, api_key):
    url = f"https://api.openweathermap.org/data/2.5/weather"
    params = {
        'q': city,
        'appid': api_key,
        'units': 'metric',
        'lang': 'ru'
    }
    
    response = requests.get(url, params=params)
    data = response.json()
    
    if response.status_code == 200:
        temp = data['main']['temp']
        print(f"Температура в {city}: {temp}°C")
        return temp
    else:
        print(f"Ошибка: {data.get('message', 'Неизвестная ошибка')}")
        return None

async def main():
    start_time = time.time()
    api_key = "13e489d16133d222e13e602f3c726e5e"
    task1 = asyncio.create_task(get_temperature("Москва", api_key))
    task2 = asyncio.create_task(get_temperature("Дубай", api_key))
    task3 = asyncio.create_task(get_temperature("Каир", api_key))
    task4 = asyncio.create_task(get_temperature("Берлин", api_key))
    task5 = asyncio.create_task(get_temperature("Пекин", api_key))

    await task1
    await task2
    await task3
    await task4
    await task5
    end_time = time.time()
    print(end_time-start_time)

final_result = await main()
```
